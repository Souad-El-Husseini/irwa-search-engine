2 - First ranking method: TF-IDF + cosine simmilarity


```python
def rank_tfidf_cosine(query: str, idx, tf, df, N:int, stops=None) -> list[tuple[str, float]]:
    """
    Rank documents using the TF-IDF cosine similarity measure.

    Parameters
    ----------
    query : str
        The query string.
    idx : dict
        The inverted index.
    tf : dict
        The term frequency dictionary.
    df : dict
        The document frequency dictionary.
    N : int
        The total number of documents in the corpus.
    stops : set, optional
        A set of stop words to ignore, by default None.

    Returns
    -------
    list[tuple[str, float]]
        A list of tuples containing document IDs and their corresponding TF-IDF cosine similarity scores,
        sorted in descending order of similarity. Only documents containing ALL query terms are returned (conjunctive search).
    """
    # Preprocess the query like we did in part 2
    query_terms = list(dict.fromkeys(preprocess(query, stops)))

    if not query_terms:
        return []

    #Perform the conjunctive search to get the candidate documents that contain all query terms
    candidates = conjunctive_search(query, idx, stops)

    if not candidates:
        return []
```
